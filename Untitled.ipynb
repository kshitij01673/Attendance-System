```python
def add_user():
    name = input("What is the name of the user you want to add: ")
    user_id, secret = db.add_user(conn,name)
    print(f"User id of user : {user_id}")
    print(f"Secret of user : {secret}")
    print("Generating QR")
    qg.generate_qr({"Name": name, "id": str(user_id), "auth_code":str(secret) })
    print("QR generated and Saved Successfully")
add_user()
```
