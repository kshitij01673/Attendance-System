```python
import cv2
import qrcode


def generate_qr(data, filename):
    """Generate a QR code."""
    img = qrcode.make(data)
    img.save(filename)
    print(f"QR code saved as {filename}")


def read_qr(filename):
    """Read a QR code from an image."""
    detector = cv2.QRCodeDetector()

    img = cv2.imread(filename)

    if img is None:
        print("Could not open image.")
        return

    data, points, _ = detector.detectAndDecode(img)

    if data:
        print("Decoded:", data)
    else:
        print("No QR code detected.")


def scan_camera():
    """Scan QR codes using the webcam."""
    detector = cv2.QRCodeDetector()
    cap = cv2.VideoCapture(0)

    if not cap.isOpened():
        print("Could not access camera.")
        return

    print("Press 'q' to quit.")

    while True:
        ret, frame = cap.read()

        if not ret:
            break

        data, points, _ = detector.detectAndDecode(frame)

        if points is not None:
            points = points.astype(int)

            # Draw green outline around QR code
            for i in range(len(points[0])):
                pt1 = tuple(points[0][i])
                pt2 = tuple(points[0][(i + 1) % len(points[0])])
                cv2.line(frame, pt1, pt2, (0, 255, 0), 2)

            if data:
                cv2.putText(
                    frame,
                    data,
                    (20, 40),
                    cv2.FONT_HERSHEY_SIMPLEX,
                    0.8,
                    (0, 255, 0),
                    2,
                )
                print("Detected:", data)

        cv2.imshow("QR Scanner", frame)

        if cv2.waitKey(1) & 0xFF == ord("q"):
            break

    cap.release()
    cv2.destroyAllWindows()


def main():
    while True:
        print("\n==== QR Tool ====")
        print("1. Generate QR Code")
        print("2. Read QR from Image")
        print("3. Scan QR with Camera")
        print("4. Exit")

        choice = input("Select: ")

        if choice == "1":
            text = input("Enter text or URL: ")
            filename = input("Save as (example qr.png): ")
            generate_qr(text, filename)

        elif choice == "2":
            filename = input("Image filename: ")
            read_qr(filename)

        elif choice == "3":
            scan_camera()

        elif choice == "4":
            break

        else:
            print("Invalid option.")


if __name__ == "__main__":
    main()
```
