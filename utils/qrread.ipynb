```python
import cv2

def scan_camera():
    """Scan a QR code using the webcam and return its contents."""
    detector = cv2.QRCodeDetector()
    cap = cv2.VideoCapture(0)

    if not cap.isOpened():
        raise RuntimeError("Could not access camera.")

    print("Show a QR code to the camera...")

    try:
        while True:
            ret, frame = cap.read()
            if not ret:
                continue

            data, points, _ = detector.detectAndDecode(frame)

            if points is not None:
                points = points.astype(int)

                # Draw green outline around the QR code
                for i in range(len(points[0])):
                    pt1 = tuple(points[0][i])
                    pt2 = tuple(points[0][(i + 1) % len(points[0])])
                    cv2.line(frame, pt1, pt2, (0, 255, 0), 2)

                if data:
                    cv2.putText(
                        frame,
                        data,
                        (20, 40),
                        cv2.FONT_HERSHEY_SIMPLEX,
                        0.8,
                        (0, 255, 0),
                        2,
                    )

                    cv2.imshow("QR Scanner", frame)
                    cv2.waitKey(500)  # Briefly display the result

                    return data

            cv2.imshow("QR Scanner", frame)

    finally:
        cap.release()
        cv2.destroyAllWindows()


# Example
content = scan_camera()
print("Scanned content:", content)
```
