# Importing a Jupyter Notebook into Another Notebook

## Option 1: Use `import-ipynb` (Simple)

Install it:

```bash
pip install import-ipynb
```

Then:

```python
import import_ipynb
import my_notebook

my_notebook.some_function()
```

This lets you import `my_notebook.ipynb` as if it were a Python module.

### Pros
- Very easy to use.
- Good for small projects.

### Cons
- Not ideal for production code.
- Some notebook-specific behavior can be unpredictable.

---

## Option 2: Convert the Notebook to a `.py` File (Recommended)

Convert the notebook:

```bash
jupyter nbconvert --to script my_notebook.ipynb
```

This creates `my_notebook.py`, which you can import normally:

```python
import my_notebook

my_notebook.some_function()
```

This is the approach most developers use because it's reliable and version-control friendly.
